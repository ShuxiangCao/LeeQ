## Single-Qubit Randomized Benchmarking

TODO: Implement and explain single-qubit RB procedures

```python
class RandomizedBenchmarkingAnalysis:
    def __init__(self, sequence_lengths, survival_probabilities, dimension=2):
        self.sequence_lengths = np.asarray(sequence_lengths)
        self.survival_probabilities = np.asarray(survival_probabilities)
        self.dimension = dimension

    def fit_decay(self):
        centered = np.clip(2 * (self.survival_probabilities - 0.5), 1e-6, 1.0)
        slope, intercept = np.polyfit(self.sequence_lengths, np.log(centered), 1)
        decay = float(np.exp(slope))
        fidelity = ((self.dimension - 1) * decay + 1) / self.dimension
        return {"decay": decay, "fidelity": fidelity, "intercept": float(intercept)}


sequence_lengths = np.array([1, 2, 4, 8, 16, 32, 64, 96])
true_decay = 0.997
survival = 0.5 + 0.5 * true_decay ** sequence_lengths
single_qubit_analysis = RandomizedBenchmarkingAnalysis(sequence_lengths, survival)
single_qubit_result = single_qubit_analysis.fit_decay()

fig = go.Figure()
fig.add_trace(go.Scatter(x=sequence_lengths, y=survival, mode="markers+lines", name="RB survival"))
fig.update_layout(title="Single-Qubit Randomized Benchmarking", xaxis_title="Clifford count", yaxis_title="Survival probability")
fig.show()

print(f"Single-qubit decay parameter: {single_qubit_result['decay']:.6f}")
print(f"Estimated average gate fidelity: {single_qubit_result['fidelity']:.6f}")
```
